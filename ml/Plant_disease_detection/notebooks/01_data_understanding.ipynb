```python
dataset_path = r"D:\Krishi-Sahayak\ml\Plant_disease_detection\dataset\plant_village"
print(dataset_path)
```

# Count Class


```python
classes = os.listdir(dataset_path)
print("Total Classes :",len(classes))
print(classes)
```

# Images per class

```python
for folder in classes:
    path = os.path.join(dataset_path, folder)

    print(folder, ":", len(os.listdir(path)))
```

# Display Images

```python
fig, axes = plt.subplots(3,3, figsize=(10,10))   #3 rows 3 columns k grid bnega matplotlib se
#fig is the image canvas like the chart paper and axes is the grid of subplots
for i, ax in enumerate(axes.flat):   #axes.flat covert krta h 2d array od axes into 1d array of axes

    folder = classes[i]  # classes is  the list of folders in the dataset path

    image = os.listdir(os.path.join(dataset_path, folder))[0]   #listdir folder k andar images ki list

    img = cv2.imread(os.path.join(dataset_path, folder, image)) 

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    ax.imshow(img)

    ax.set_title(folder)

    ax.axis("off")

plt.tight_layout()

plt.show()
```
